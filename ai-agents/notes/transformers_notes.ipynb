НЛП — это область лингвистики и машинного обучения, сосредоточенная на понимании всего, что связано с человеческим языком. Цель заданий по НЛП — не только понять отдельные слова по отдельности, но и уметь понимать их контекст.

Ниже приведён список распространённых задач NLP с примерами каждой:

Генерация текстового контента: заполнение запроса с автосгенерированным текстом, заполнение пробелов в тексте замаскированными словами

Рост крупных языковых моделей (LLM)
В последние годы область НЛП была революционизирована Крупными языковыми моделями (LLM). Эти модели, включающие такие архитектуры, как GPT (Generative Pre-trained Transformer) и Llama, изменили возможности в обработке языка.

Большая языковая модель (LLM) — это модель ИИ, обученная на огромных объёмах текстовых данных, способная понимать и генерировать текст, похожий на человек, распознавать языковые шаблоны и выполнять широкий спектр языковых задач без обучения, связанного с конкретными задачами. Они представляют собой значительный прогресс в области обработки естественного языка (NLP).

LLM характеризуются:

Масштаб: Они содержат миллионы, миллиарды или даже сотни миллиардов параметров
Общие способности: Они могут выполнять несколько задач без обучения конкретным задачам
Обучение в контексте: Они могут учиться на примерах, приведённых в задании
Эмерджентные способности: по мере роста размера этих моделей они демонстрируют возможности, которые не были явно запрограммированы или предвидены
Появление LLM изменило парадигму от построения специализированных моделей для конкретных задач NLP на использование одной крупной модели, которую можно подсказывать или тонко настраивать для решения широкого спектра языковых задач. Это сделало сложную обработку языка более доступной, а также создало новые вызовы в таких областях, как эффективность, этика и внедрение.

Однако у LLM есть и важные ограничения:

Галлюцинации: они могут уверенно выдать неверную информацию
Отсутствие истинного понимания: они не обладают истинным пониманием мира и действуют исключительно на статистических закономерностях
Смещение: Они могут воспроизводить предвзятости, присутствующие в их обучающих данных или входных данных.
Контекстные окна: У них ограниченное количество контекстных окон (хотя это улучшается)
Вычислительные ресурсы: они требуют значительных вычислительных ресурсов
Почему обработка языка сложна?
Компьютеры не обрабатывают информацию так же, как люди. Например, когда мы читаем предложение «Я голоден», мы легко понимаем его значение. Аналогично, имея два предложения, такие как «Я голоден» и «Мне грустно», мы легко можем определить, насколько они похожи. Для моделей машинного обучения (ML) такие задачи сложнее. Текст должен быть обработан так, чтобы модель могла извлечь из него уроки. И поскольку язык сложен, нам нужно тщательно подумать, как должна осуществляться эта обработка. Проведено много исследований по представлению текста, и мы рассмотрим некоторые методы в следующей главе.

Даже с развитием LLM остаётся множество фундаментальных вызовов. К ним относятся понимание неоднозначности, культурного контекста, сарказма и юмора. LLM решают эти задачи через масштабное обучение на различных наборах данных, но при этом часто не дотягивают до человеческого понимания во многих сложных ситуациях.

Самый базовый объект в библиотеке 🤗 Transformers — это функция. Он связывает модель с необходимыми этапами предварительной и послеобработки, позволяя нам напрямую вводить любой текст и получать понятный ответ:pipeline()

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
classifier("I've been waiting for a HuggingFace course my whole life.")

[{'label': 'POSITIVE', 'score': 0.9598047137260437}]
```

Мы даже можем вынести несколько приговоров!

```python
classifier(
    ["I've been waiting for a HuggingFace course my whole life.", "I hate this so much!"]
)
```

```python
[{'label': 'POSITIVE', 'score': 0.9598047137260437},
 {'label': 'NEGATIVE', 'score': 0.9994558095932007}]
```

По умолчанию этот конвейер выбирает определённую предобученную модель, которая была доработана для анализа настроений на английском языке. Модель загружается и кэшируется при создании объекта. Если вы перезапустите команду, вместо неё будет использована кэшированная модель, и скачивать её снова не нужно.classifier

Классификация нулевого выстрела

Классификация нулевого выстрела
Начнём с более сложной задачи — классифицировать тексты, которые не были подписаны. Это распространённая ситуация в реальных проектах, поскольку аннотирование текста обычно занимает много времени и требует экспертизы в области. В этом случае конвейер очень мощен: он позволяет указать, какие метки использовать для классификации, так что не нужно полагаться на метки предварительно обученной модели. Вы уже видели, как модель может классифицировать предложение как положительное или отрицательное с помощью этих двух ярлыков — но также может классифицировать текст с помощью любого другого набора ярлыков, которые вам нравятся.zero-shot-classification

```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification")
classifier(
    "This is a course about the Transformers library",
    candidate_labels=["education", "politics", "business"],
)
```

```python
{'sequence': 'This is a course about the Transformers library',
 'labels': ['education', 'business', 'politics'],
 'scores': [0.8445963859558105, 0.111976258456707, 0.043427448719739914]}
```

Этот конвейер называется zero-shot, потому что не нужно тонко настраивать модель данных для её использования. Он может напрямую возвращать вероятностные оценки для любого списка меток, который вы хотите!

Генерация текста
Теперь давайте посмотрим, как использовать пайплайн для генерации текста. Основная идея здесь в том, что вы даёте подсказку, и модель автоматически заполняет его, генерируя оставшийся текст. Это похоже на функцию предиктивного текста, которая используется на многих телефонах. Генерация текста связана с случайностью, так что это нормально, если вы не получаете тех же результатов, что показано ниже.

```python
from transformers import pipeline

generator = pipeline("text-generation")
generator("In this course, we will teach you how to")
```
