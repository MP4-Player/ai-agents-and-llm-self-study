```python
from openai import OpenAI
```

```python
API_KEY="ollama"
BASE_URL="http://localhost:11434/v1"
MODEL_NAME="gemma3:4b"
```

```python
def generate_answer(system_prompt: str, user_prompt: str):

    client= OpenAI(api_key=API_KEY,base_url=BASE_URL)
    response = client.chat.completions.create(
        model=MODEL_NAME,
        messages=[
            {"role":"system","content": system_prompt},
            {"role":"system","content":user_prompt}
        ]
    )
    answer=response.choices[0].message.content
    print(f"AI: {answer}") 
    
    # реализуйте вызов LLM (используйте пример выше)
    pass
```

```python
system_prompt = "Ты ИИ асистент который учит писать ИИ асистентов . Отвечаешь на русском. Без эмоджи" # напишите системную инструкцию
user_prompt = "Приведи 3 популярных ошибки начинающих программистов в создании ИИ асистента ."
cnt = 2

while cnt > 0:
    user_prompt = input(">")
    print(f"Вы: {user_prompt}") 
    answer = generate_answer(system_prompt, user_prompt)
    
    #print(f"ИИ:\n{answer}\n")

    cnt -= 1
```
